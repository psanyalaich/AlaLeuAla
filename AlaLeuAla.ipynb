```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

phi = np.random.normal(loc=-60, scale=40, size=3000)  # Beta & alpha region
psi = np.random.normal(loc=45, scale=40, size=3000)   # Beta & alpha region

phi_value = -86.4526
psi_value = 143.018

plt.figure(figsize=(8,6))
sns.kdeplot(x=phi, y=psi, cmap="mako", fill=True, thresh=0.05, levels=50)
plt.scatter(phi_value, psi_value, color='red', s=100, edgecolors='black', label="Leu (Residue 2)")

plt.xlabel("Phi (φ) Angle (°)")
plt.ylabel("Psi (ψ) Angle (°)")
plt.title("Ramachandran Plot for Ala-Leu-Ala")
plt.xlim(-180, 180)
plt.ylim(-180, 180)
plt.grid(True, linestyle="--", linewidth=0.5)
plt.legend()
plt.show()
```
